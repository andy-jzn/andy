```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data=np.loadtxt(r"C:\Users\andyj\OneDrive\Escritorio\python_decal_fa25\andy\homework7\star_data.txt", skiprows=1)
```

```python
print("Data shape:", data.shape)
print("Number of rows (stars):", data.shape[0])
```

```python
#rows = 2642
```

```python
magnitude = data[:, 0]
temperature=data[:, 1]
```

```python
print(f"Minimum magnitude (brightest): {np.min(magnitude)}")
print(f"Maximum magnitude (dimmest): {np.max(magnitude)}")
print(f"Minimum temperature (coldest): {np.min(temperature)}")
print(f"Maximum temperature (hottest): {np.max(temperature)}")
```

```python
brightest_idx = np.argmin(magnitude)
dimmest_idx   = np.argmax(magnitude)
hottest_idx   = np.argmax(temperature)
coldest_idx   = np.argmin(temperature)
```

```python
log_magnitude = np.log(magnitude - np.min(magnitude) + 1)  # shift to avoid log(0)
log_temperature = np.log(temperature)
```

```python
plt.figure(figsize=(8, 6))
sc = plt.scatter(log_temperature, log_magnitude, c=temperature, cmap='coolwarm', label='Stars')

#plt.gca().invert_xaxis()
#plt.gca().invert_yaxis()

plt.scatter(log_temperature[brightest_idx], log_magnitude[brightest_idx], color='black', marker='*', s=300, label='Brightest Star')
plt.scatter(log_temperature[dimmest_idx], log_magnitude[dimmest_idx], color='purple', marker='*', s=300, label='Dimmest Star')
plt.scatter(log_temperature[hottest_idx], log_magnitude[hottest_idx], color='green', marker='*', s=300, label='Hottest Star')
plt.scatter(log_temperature[coldest_idx], log_magnitude[coldest_idx], color='orange', marker='*', s=300, label='Coldest Star')

plt.title("Hertzsprung–Russell Diagram", fontsize=16)
plt.xlabel("Temperature (log K)", fontsize=14)
plt.ylabel("Magnitude (log scale)", fontsize=14)
plt.legend(fontsize=10)

cbar= plt.colorbar(sc)
cbar.set_label("Temperature (K)")

plt.show()
```
