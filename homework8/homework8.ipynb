```python
median_lat=df["latitude"].median()
median_lon=df["longitude"].median()

print(f"Median Latitude: ", median_lat)
print(f"Median Longiitude: ", median_lon)

df["dist_to_median_lat"] = (df["latitude"] - median_lat).abs()
df["dist_to_median_lon"] = (df["longitude"] - median_lon).abs()


print("New DataFrame shape:", df.shape)
```

```python
##There are 52 rows and 6 columns
```

```python
# Closest and farthest by latitude
closest_lat_state = df.loc[df["dist_to_median_lat"].idxmin()]
farthest_lat_state = df.loc[df["dist_to_median_lat"].idxmax()]

# Closest and farthest by longitude
closest_lon_state = df.loc[df["dist_to_median_lon"].idxmin()]
farthest_lon_state = df.loc[df["dist_to_median_lon"].idxmax()]

print(f"Closest to median latitude: {closest_lat_state['state']} ({closest_lat_state['dist_to_median_lat']:.2f})")
print(f"Farthest from median latitude: {farthest_lat_state['state']} ({farthest_lat_state['dist_to_median_lat']:.2f})")
print(f"Closest to median longitude: {closest_lon_state['state']} ({closest_lon_state['dist_to_median_lon']:.2f})")
print(f"Farthest from median longitude: {farthest_lon_state['state']} ({farthest_lon_state['dist_to_median_lon']:.2f})")
```

```python
df = sns.load_dataset("penguins")

print(df.shape)
print(df.head())
print(df.tail())
```

```python
"""
a) it has 344 columns, and 7 rows
b) columns:
species: penguin species
island: the island in the Palmer Archipelago
bill_lenghth: bill (beak) length in millimeters
bill_depth_mm: bill depth in milimeters
flipper_length_mm: bill depth in millimiters
body_mass_g: body mass in grams
sex: male or female
"""
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Top-left: Scatter plot (bill length vs. flipper length)
sns.scatterplot(x="bill_length_mm", y="flipper_length_mm", data=df, hue="species", ax=axes[0, 0])
axes[0, 0].set_title("Bill Length vs. Flipper Length")
axes[0, 0].set_xlabel("Bill Length (mm)")
axes[0, 0].set_ylabel("Flipper Length (mm)")
axes[0, 0].legend(title="Species")

# Top-right: Scatter plot (bill depth vs. body mass)
sns.scatterplot(x="bill_depth_mm", y="body_mass_g", data=df, hue="sex", ax=axes[0, 1])
axes[0, 1].set_title("Bill Depth vs. Body Mass")
axes[0, 1].set_xlabel("Bill Depth (mm)")
axes[0, 1].set_ylabel("Body Mass (g)")
axes[0, 1].legend(title="Sex")

# Bottom-left: Histogram (body mass distribution)
sns.histplot(x="body_mass_g", data=df, bins=20, color="skyblue", ax=axes[1, 0])
axes[1, 0].set_title("Body Mass Distribution")
axes[1, 0].set_xlabel("Body Mass (g)")
axes[1, 0].set_ylabel("Count")

# Bottom-right: Boxplot (flipper length by species)
sns.boxplot(x="species", y="flipper_length_mm", data=df, palette="pastel", ax=axes[1, 1])
axes[1, 1].set_title("Flipper Length by Species")
axes[1, 1].set_xlabel("Species")
axes[1, 1].set_ylabel("Flipper Length (mm)")

plt.tight_layout()
plt.show()
```
