```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= pd.read_csv("exoplanet_data.csv", comment="#")
print(df.shape)
```

```python
print(f"Shape of dataset: {df.shape}")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
print("\nColumn titles:")
print(df.columns.tolist())
```

```python
"""
a) shape (567, 92)
b) 567 rows, 92 columns
c) pl_orbper, st_teff, sy_gaiamag, dy_dist, pl_insol
"""
```

```python
#reducing data 4.3
```

```python
"""
a) Planet Name: pl_name – The official planet designation
b) Host Name: hostname – The star that the planet orbits
c) Discovery Method: discoverymethod – Detection technique
d) Spectral Type: st_spectype – Classification of the host star
e) Equilibrium Temperature [K]: pl_eqt – Estimated temperature assuming radiative equilibrium
f) Planet Mass [Jupiter Mass]: pl_bmassj – Planet’s mass in Jupiter masses
"""
```

```python
df_small = df[["pl_name", "hostname", "discoverymethod", "st_spectype", "pl_eqt", "pl_bmassj"]]
df_small = df_small.dropna()
df_small.shape
```

```python

method_counts = df_small["discoverymethod"].value_counts()

print(method_counts)
print("\nMost common discovery method:", method_counts.idxmax())
print("Least common discovery method:", method_counts.idxmin())
```

```python
"""
a) discovery method: transit, radial velocity and imaging
b) Most common, transit, least common is imaging
c) Transit: Detects dips in star brightness as a planet crosses in front.
    Radial Velocity: Detects star’s “wobble” caused by gravitational pull of orbiting planet.
    Imaging: Takes direct photos of the planet using telescopes.
"""
```

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df_small,
    x="pl_bmassj",
    y="pl_eqt",
    hue="discoverymethod",
    palette="viridis"
)

plt.title("Exoplanet Mass vs. Equilibrium Temperature")
plt.xlabel("Planet Mass (Jupiter Masses)")
plt.ylabel("Equilibrium Temperature (K)")
plt.legend(title="Discovery Method")
plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.scatterplot(
    x=np.log(df_small["pl_bmassj"]),
    y=np.log(df_small["pl_eqt"]),
    hue=df_small["discoverymethod"],
    palette="viridis"
)

plt.title("Log-Scaled Exoplanet Mass vs. Temperature")
plt.xlabel("log(Planet Mass)")
plt.ylabel("log(Equilibrium Temperature)")
plt.legend(title="Discovery Method")
plt.show()
```

```python
#exoplanets found by imaging are more likely to be massive and colder
```

```python
cold_massive = df_small[df_small["pl_bmassj"] > 5].sort_values(by="pl_eqt").iloc[0]
print(cold_massive)
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(
    x=np.log(df_small["pl_bmassj"]),
    y=np.log(df_small["pl_eqt"]),
    hue=df_small["discoverymethod"],
    palette="coolwarm"
)

# Highlight coldest massive planet
plt.scatter(
    np.log(cold_massive["pl_bmassj"]),
    np.log(cold_massive["pl_eqt"]),
    color="black",
    s=250,
    marker="*",
    label=cold_massive["pl_name"]
)

plt.title("Exoplanet Mass vs. Temperature (Log Scale)")
plt.xlabel("log(Planet Mass)")
plt.ylabel("log(Equilibrium Temperature)")
plt.legend()
plt.show()
```
