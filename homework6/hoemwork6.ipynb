```python
#4 Prime Star Systems
```

```python
arr= np.array([
    [2, 3, 5],
    [14, 6, 8],
    [11, 13, 17],
    [7, 10, 13]
])

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

def rows_with_primes(matrix):
    mask = np.array([any(is_prime(x) for x in row) for row in matrix])
    return matrix[mask]
    
result = rows_with_primes(arr)
print(result)
```

```python
##Checkers!
```

```python
def checkerboard():
    board = np.zeros((8, 8), dtype=int)
    board[::2, ::2] = 1   
    board[1::2, 1::2] = 1 
    return board

print(checkerboard())
```

```python
def reverse_checkerboard():
    board = np.zeros((8, 8), dtype=int)
    board[::2, 1::2] = 1  
    board[1::2, ::2] = 1   
    return board
print(reverse_checkerboard())
```

```python
#The Expanding Universe
```

```python
def expansion(arr,n):
    spacer= ' ' *n
    expanded =[spacer.join(word) for word in arr]
    return np.array(expanded)

universe= np.array(['galaxy', 'cluster'])
print(expansion(universe, 1))
print(expansion(universe, 2))
```

```python
##Second- Brightest Star
```

```python
def second_brightest(stars):
    sorted_rows=np.sort(stars, axis=1)
    return sorted_rows[:, -2]

#EG.
np.random.seed(123)
stars = np.random.randint(500, 2000, (5, 5))
print(stars)
print(second_brightest(stars))
```
