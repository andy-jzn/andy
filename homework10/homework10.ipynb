```python

skip_cols = ["Planet", "Ring System?", "Global Magnetic Field?"]


for col in df.columns:
    if col not in skip_cols:
        df[col] = pd.to_numeric(df[col], errors="coerce")

print(df.dtypes)
```

```python
from astropy import units as u

def attach_units(col):
    
    if col == "Mass (10^24kg)":
        df[col] = df[col] * 1e24 * u.kg
        df.rename(columns={col: "Mass (kg)"}, inplace=True)

    elif col == "Radius (km)":
        df[col] = df[col] * u.km
        df.rename(columns={col: "Radius (km) [km]"}, inplace=True)

    elif col == "Orbital Period (days)":
        df[col] = df[col] * u.day
        df.rename(columns={col: "Orbital Period (days) [day]"}, inplace=True)

    elif col == "Equilibrium Temperature (K)":
        df[col] = df[col] * u.K
        df.rename(columns={col: "Equilibrium Temperature (K) [K]"}, inplace=True)

    elif col == "Distance (km)":
        df[col] = df[col] * u.km
        df.rename(columns={col: "Distance (km) [km]"}, inplace=True)

        
for col in df.columns:
    if col not in ["Planet", "Ring System?", "Global Magnetic Field?"]:
        attach_units(col)
```

```python

df["Semi-Major Axis (km)"] = (df["Perihelion (10^6 km)"] + df["Aphelion (10^6 km)"]) / 2 * 1e6 * u.km

print(df["Semi-Major Axis (km)"].head())

print("Columns after inserting Semi-Major Axis:")
print(df.columns)
print("\nSemi-Major Axis sample:\n", df["Semi-Major Axis (km)"].head())
```

```python
period_col = "Orbital Period (days) [day]"


df[period_col] = df[period_col].astype(float)

df["Orbital Period (years)"] = (df[period_col].values * u.day).to(u.yr)

print("Updated columns:")
print(df.columns, "\n")

print("Orbital Period (years) sample:")
print(df["Orbital Period (years)"].head(), "\n")


earth_period = df.loc[df["Planet"] == "Earth", "Orbital Period (years)"].values[0]
print(f"Earth's orbital period: {earth_period:.4f}")
```

```python

print("1 AU =", const.au)
print("In kilometers =", const.au.to(u.km))
```

```python
"""
AU means the distance between the Earth and the Sun
there are 149597870 km in one AU
"""
```

```python
distance_cols_to_convert = [
    "Perihelion (10^6 km)",
    "Aphelion (10^6 km)",
    "Distance from Sun (10^6 km)",
]


for col in distance_cols_to_convert:
    df[col] = df[col].astype(float).values
    df[col] = (df[col].values * u.km).to_value(u.au)
    new_col_name = col.replace("(10^6 km)", "(AU)")
    df.rename(columns={col: new_col_name}, inplace=True)
```
